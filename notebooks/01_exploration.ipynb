## 4. Daily Returns

```python
df["daily_return"] = df["close"].pct_change()

fig = px.line(
    df,
    x="date",
    y="daily_return",
    title="Daily Returns - RELIANCE"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Date",
    yaxis_title="Daily Return"
)

fig.show()
```

## 5. Cumulative Returns

```python
df["cumulative_return"] = (1 + df["daily_return"]).cumprod()

fig = px.line(
    df,
    x="date",
    y="cumulative_return",
    title="Cumulative Returns"
)

fig.update_layout(
    template="plotly_white"
)

fig.show()
```

## 6. Maximum Drawdown Analysis

```python
rolling_max = df["close"].cummax()

df["drawdown"] = (
    df["close"] - rolling_max
) / rolling_max

fig = px.area(
    df,
    x="date",
    y="drawdown",
    title="Portfolio Drawdown"
)

fig.update_layout(
    template="plotly_white"
)

fig.show()
```
